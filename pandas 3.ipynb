```python
import pandas as pd

data = {
    "Name": ["Asha", "Rahul", "Priya", "Kiran", "Sneha"],
    "Marks": [85, 60, 92, 45, 78],
    "Age": [20, 21, 19, 22, 20]
}

df = pd.DataFrame(data)

print(df)
```

```python
high_marks = df[df["Marks"] > 75]

print(high_marks)
```

```python
low_marks = df[df["Marks"] < 70]

print(low_marks)
```

```python
filtered_students = df[(df["Marks"] > 75) & (df["Age"] < 21)]

print(filtered_students)
```

```python
print("Number of students:", len(filtered_students))
```
